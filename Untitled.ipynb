```python
import geopandas as gpd

farm = gpd.read_file("nyathuna_farm.kml", driver="KML")
# Diagnostics
# print(farm)
# print(farm.geometry)
```

```python
farm_utm = farm.to_crs("EPSG:32737")
# Diagnostics
# print(farm_utm[["Name", "geometry"]].to_string())
# print(farm_utm.total_bounds)
```

```python
SENTINEL_DATA_PATH_10M = "s2c_nyathuna/GRANULE/L2A_T37MBU_A010787_20260929T075741/IMG_DATA/R10m/"
SENTINEL_BO4_PATH = f"{SENTINEL_DATA_PATH_10M}/T37MBU_20260929T073711_B04_10m.jp2" # RED
SENTINEL_BO8_PATH = f"{SENTINEL_DATA_PATH_10M}/T37MBU_20260929T073711_B08_10m.jp2" # NIR
SENTINEL_BO2_PATH = f"{SENTINEL_DATA_PATH_10M}/T37MBU_20260929T073711_B02_10m.jp2" # BLUE
SENTINEL_BO3_PATH = f"{SENTINEL_DATA_PATH_10M}/T37MBU_20260929T073711_B03_10m.jp2" # GREEN
```

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from rasterio.features import geometry_window

with rasterio.open(SENTINEL_BO4_PATH) as red_src, \
     rasterio.open(SENTINEL_BO8_PATH) as nir_src, \
     rasterio.open(SENTINEL_BO3_PATH) as green_src, \
     rasterio.open(SENTINEL_BO2_PATH) as blue_src:

    # Same spatial window for all bands
    window = geometry_window(
        red_src,
        farm_utm.geometry.tolist()
    )

    # Read bands
    red = red_src.read(1, window=window).astype(float)
    green = green_src.read(1, window=window).astype(float)
    blue = blue_src.read(1, window=window).astype(float)
    nir = nir_src.read(1, window=window).astype(float)

    # Spatial transform
    transform = red_src.window_transform(window)

# NDVI
ndvi = (nir - red) / (nir + red)
```

```python
import rasterio.plot
```

```python
# Stack into RGB order
rgb = np.stack([red, green, blue], axis=-1)

# Convert reflectance scaling
rgb = rgb / 10000.0

# Stretch RGB for visualization
rgb_display = np.clip(rgb / 0.3, 0, 1)
```

```python
fig, ax = plt.subplots(figsize=(9, 8))

# RGB background
ax.imshow(
    rgb_display,
    extent=(
        transform.c,
        transform.c + transform.a * rgb.shape[1],
        transform.f + transform.e * rgb.shape[0],
        transform.f
    )
)

# NDVI overlay
ndvi_img = ax.imshow(
    ndvi,
    extent=(
        transform.c,
        transform.c + transform.a * ndvi.shape[1],
        transform.f + transform.e * ndvi.shape[0],
        transform.f
    ),
    cmap="RdYlGn",
    vmin=0,
    vmax=0.4,
    alpha=0.45
)

# Farm boundaries
farm_utm.boundary.plot(
    ax=ax,
    linewidth=2
)

# Farm labels
for _, row in farm_utm.iterrows():

    point = row.geometry.representative_point()

    ax.text(
        point.x,
        point.y,
        row["Name"],
        fontsize=10,
        fontweight="bold",
        ha="center",
        va="center"
    )

# Colorbar
cbar = plt.colorbar(ndvi_img, ax=ax)
cbar.set_label("NDVI")

ax.set_xlabel("Easting (UTM metres)")
ax.set_ylabel("Northing (UTM metres)")
ax.set_title("Sentinel-2 True Color with NDVI Overlay")

ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```

# NDVI PLOT

```python
import rasterio
import matplotlib.pyplot as plt
import geopandas as gpd
from rasterio.features import geometry_window

with rasterio.open(SENTINEL_BO4_PATH) as red_src, \
     rasterio.open(SENTINEL_BO8_PATH) as nir_src:

    # Window containing the four farms
    window = geometry_window(
        red_src,
        farm_utm.geometry.tolist()
    )

    # Read Red and NIR
    red = red_src.read(1, window=window).astype(float)
    nir = nir_src.read(1, window=window).astype(float)

    # NDVI
    ndvi = (nir - red) / (nir + red)

    # Spatial transform of the window
    transform = red_src.window_transform(window)


fig, ax = plt.subplots(figsize=(9, 8))

# ---------------------------------------------------------
# NDVI
# ---------------------------------------------------------

img = ax.imshow(
    ndvi,
    extent=(
        transform.c,
        transform.c + transform.a * ndvi.shape[1],
        transform.f + transform.e * ndvi.shape[0],
        transform.f
    ),
    cmap="RdYlGn",
    vmin=0,
    vmax=0.4
)


# ---------------------------------------------------------
# Farm boundaries
# ---------------------------------------------------------

farm_utm.boundary.plot(
    ax=ax,
    linewidth=2
)


# ---------------------------------------------------------
# Farm labels
# ---------------------------------------------------------

for _, row in farm_utm.iterrows():

    label_point = row.geometry.representative_point()

    ax.text(
        label_point.x,
        label_point.y,
        row["Name"],
        fontsize=10,
        fontweight="bold",
        ha="center",
        va="center"
    )


# ---------------------------------------------------------
# Actual Sentinel-2 pixel boundaries
# ---------------------------------------------------------

pixel_size = abs(transform.a)

left = transform.c
right = left + pixel_size * ndvi.shape[1]

top = transform.f
bottom = top - pixel_size * ndvi.shape[0]

# Vertical boundaries
for x in range(
    int(left),
    int(right) + 1,
    int(pixel_size)
):
    ax.axvline(
        x,
        linewidth=0.7,
        alpha=0.5
    )

# Horizontal boundaries
for y in range(
    int(bottom),
    int(top) + 1,
    int(pixel_size)
):
    ax.axhline(
        y,
        linewidth=0.7,
        alpha=0.5
    )


# ---------------------------------------------------------
# Formatting
# ---------------------------------------------------------

cbar = plt.colorbar(img, ax=ax)
cbar.set_label("NDVI")

ax.set_xlabel("Easting (UTM metres)")
ax.set_ylabel("Northing (UTM metres)")

ax.set_title(
    "Sentinel-2 NDVI with Experimental Farm Plots"
)

ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```
